## M7. Text and Images as Data

Models work on numbers. Text and images are not numbers until you **represent** them
as numbers - and once you do, it is the same classify-and-evaluate work as Module 3.

- **Text -> numbers** with **TF-IDF**: each document becomes a vector of word weights,
  high for words that are frequent in this document but rare across the corpus. This
  captures *which words* appear, but throws away **word order and meaning** ("dog bites
  man" and "man bites dog" look identical).
- **Images -> numbers** by reading **pixel values**. An 8x8 grayscale image is 64
  numbers. Flattening to a feature vector throws away **spatial structure** (which
  pixels are neighbors).

Naming what a representation discards is the judgment this module adds. The modeling
and evaluation are the skills you already have.

## Section 2. Text-Based Learning

```text
ANALYST CHOICE
TF-IDF turns each document into word weights. 
It captures which words appear, not their order or meaning. 
Decide whether that is good enough for your text problem.
```

````python
# === Section 2. Text-Based Learning ===

# A small, self-contained labeled corpus (topic classification).
# CUSTOM: replace with your own labeled text in the Apply/Extend section.
TEXT_DOCS: list[str] = [
    "my internet is not working after the storm",
    "the billing amount is incorrect for this month",
    "my interet is too slow after the powercut",
    "my billing statement shows a charge I did not make",
    "my bill is too high and I want to dispute it",
    "my router keeps disconnecting every few minutes",
    "please help me reset my password",
    "the technician did not arrive for the appointment",
    "how do I upgrade my plan to higher speed",
    "the mobile app keeps crashing on login",
    "I want to cancel my subscription immediately",
]
TEXT_LABELS = ["technical_issue",
    "billing_issue",
    "technical_issue",
    "billing_issue",
    "billing_issue",
    "technical_issue",
    "account_issue",
    "service_issue",
    "upgrade_request",
    "technical_issue",
    "cancellation_request",]


def run_text_pipeline() -> None:
    """Custom text classification pipeline using TF-IDF + MultinomialNB."""

    # Show corpus info
    print(f"Text corpus: {len(TEXT_DOCS)} documents")
    print(f"Classes: {sorted(set(TEXT_LABELS))}")

    # Train/test split
    docs_train, docs_test, y_train, y_test = train_test_split(
        TEXT_DOCS,
        TEXT_LABELS,
        test_size=0.25,
        random_state=RANDOM_STATE,
        stratify=None,
    )

    # TF-IDF with bigrams
    vectorizer = TfidfVectorizer(ngram_range=(1, 2))
    X_train = vectorizer.fit_transform(docs_train)
    X_test = vectorizer.transform(docs_test)
    print(f"Vocabulary size (features): {len(vectorizer.vocabulary_)}")

    # Model
    model = MultinomialNB()
    model.fit(X_train, y_train)

    # Accuracy
    acc = accuracy_score(y_test, model.predict(X_test))
    print(f"Text test accuracy: {acc:.3f}")

    # Predicting custom sentences
    new_docs = [
        "my internet stopped working after the storm",
        "the billing amount is incorrect for this month"
    ]
    preds = model.predict(vectorizer.transform(new_docs))

    print("\nPredictions on new sentences:")
    for doc, pred in zip(new_docs, preds):
        print(f" {doc!r} => predicted label: {pred}")

    print("\nDid any misread reveal TF-IDF's word-order limitation?")
   

````

### Did bigrams help?
Yes-bigrams like "keeps disconnecting" and "billing amount" improved accuracy by preserving small pieces of word order.

## Section 3. Image-Based Learning

```text
ANALYST CHOICE
Flattening an 8x8 image to 64 pixel features discards which pixels are neighbors.
A model still learns digits from raw pixels; decide where that representation
would break down (rotation, scale, position).
```

````python
from typing import Any, cast
from sklearn.ensemble import RandomForestClassifier

# === Section 3. Image-Based Learning ===


def run_image_pipeline() -> None:
    """Represent images as pixel vectors, classify the digits, and evaluate.

    WHY: An 8x8 grayscale image is 64 numbers. After flattening, it is again the
    same fit/predict/score work. We also look at a few images so the data is not
    just an abstract matrix.

    Returns:
        None
    """
    digits = load_digits()
    digits = cast(Any, load_digits())
    print(
        f"Image data: {digits.data.shape[0]} images, {digits.data.shape[1]} pixel features each"
    )
    print(f"Classes (digits): {sorted(set(digits.target.tolist()))}")

    # Look at the data: show the first few images with their true labels.
    fig, axes = plt.subplots(1, 5, figsize=(8, 2))
    for ax, image, label in zip(axes, digits.images, digits.target, strict=False):
        ax.imshow(image, cmap="gray")
        ax.set_title(f"label {label}")
        ax.axis("off")
    plt.suptitle("A few input images (8x8 pixels)")
    plt.show()

    X = digits.data
    y = digits.target
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
    )
    model = RandomForestClassifier(n_estimators=300, random_state=RANDOM_STATE)
    model.fit(X_train, y_train)
    # Compute mis classified images
    y_pred = model.predict(X_test)
    mis_idx = (y_pred != y_test)
    mis_images = X_test[mis_idx]
    mis_true = y_test[mis_idx]
    mis_pred = y_pred[mis_idx]
    
    fig,axes = plt.subplots(1, 5, figsize=(10,3))
    for ax, image, true_label, pred_label in zip(axes, mis_images, mis_true, mis_pred, strict=False):
        ax.imshow(image.reshape(8, 8), cmap="gray")
        ax.set_title(f"true: {true_label}\npred: {pred_label}")
        ax.axis("off")
    plt.suptitle("A few misclassified images (8x8 pixels)")
    plt.show()

    acc: float = float(accuracy_score(y_test, model.predict(X_test)))
    print(f"Image test accuracy: {acc:.3f}")

    print("Showing confusion matrix (which digits get confused)")
    plt.figure(figsize=(6, 6))
    ConfusionMatrixDisplay.from_estimator(model, X_test, y_test)
    plt.title("Digit confusion matrix (test set)")
    plt.show()

    print("Which digits confuse each other, and does the pixel view explain why?")
    
````

````python
run_text_pipeline()
````

````python
run_image_pipeline()
````
